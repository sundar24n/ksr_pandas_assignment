```python
import pandas as pd

sales = pd.DataFrame({
    "order_id": ["O001", "O002", "O003", "O003", "O005", "O006"],
    "product": ["Laptop", "Mouse", "Keyboard", "Keyboard", "Monitor", "Laptop"],
    "quantity": ["2", "5", "0", "-1", "3", "4"],
    "unit_price": ["50000", "500", "0", "-100", "15000", "55000"],
    "payment_method": [
        "UPI",
        "upi",
        "Credit Card",
        "credit card",
        "COD",
        None
    ],
    "order_date": [
        "2025-01-10",
        "10-02-2025",
        "2025/03/15",
        "2025/03/15",
        "2025-04-20",
        "2025-99-99"
    ]
})
```
