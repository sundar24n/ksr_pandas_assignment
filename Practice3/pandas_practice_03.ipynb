```python
import pandas as pd

customer = pd.DataFrame({
    "customer_id": ["C001", "C002", "C003", "C004", "C005"],
    "name": ["Ravi", "Priya", "Kumar", "Ravi", "Anita"],
    "age": ["35", "Unknown", "?", "-5", "53.0"],
    "gender": ["Male", "female", "M", "MALE", "F"],
    "city": ["Hyderabad", "Chennai", "Hyderabad", "hyderabad", None]
})

customer
```

```python
#1. age contains Unknown, ?, negative value and "53.0"

#convert age into number format

customer["age"]=pd.to_numeric(customer["age"],errors="coerce").astype("Int64")
customer.loc[~customer["age"].between(0,100),"age"]=pd.NA
customer
```

```python
#2.Gender has inconsistent values.


customer["gender"] = customer["gender"].replace({
    "M": "Male",
    "MALE": "Male",
    "F": "Female",
    "female": "Female"
})

customer
```

```python

#3.City has inconsistent capitalization.

customer["city"]=customer["city"].str.title()
customer

#4. Missing city.

customer["city"]=customer["city"].fillna("unknown")
customer
```

```python
#5. Duplicate customer name exists, but customer_id is unique.

customer.drop_duplicates(subset=["name","age","gender","city"])
```
